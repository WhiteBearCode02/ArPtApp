```python
import os
import zipfile
from google.colab import drive

# [기능 1] 클라우드 스토리지 안전 마운트
# - 기존 마운트가 꼬였을 상황을 대비하여 강제 리마운트 옵션 부여
drive.mount('/content/drive', force_remount=True)

# [기능 2] Source(입력) 경로 설정: 구글 드라이브 (네트워크 통신)
# - 무거운 zip 파일은 안전한 클라우드 드라이브에 두고 스트리밍 방식으로 읽어옴
drive_base_dir = '/content/drive/MyDrive/AI_Project'
img_zip_path = os.path.join(drive_base_dir, '[원천]validation_필기체.zip')
label_zip_path = os.path.join(drive_base_dir, '[라벨]validation_필기체.zip')

# [기능 3] Destination(출력) 경로 수정: 코랩 초고속 로컬 디스크 (NVMe SSD)
# - 수만 개의 파일 쓰기(Write) 작업을 네트워크가 아닌 로컬 메모리에서 처리하도록 경로 변경
local_base_dir = '/content/AIHUB_Raw'
img_extract_path = os.path.join(local_base_dir, 'images')
label_extract_path = os.path.join(local_base_dir, 'labels')

# [기능 4] 타겟 디렉토리 초기화 및 생성
os.makedirs(img_extract_path, exist_ok=True)
os.makedirs(label_extract_path, exist_ok=True)

# [기능 5] 원천 데이터 로컬 압축 해제 파이프라인
print("🚀 초고속 로컬 디스크로 원천 데이터 압축 해제 중... (수 초 ~ 1분 내외 예상)")
if os.path.exists(img_zip_path):
    with zipfile.ZipFile(img_zip_path, 'r') as zip_ref:
        zip_ref.extractall(img_extract_path) # 로컬 경로에 수만 장 쏟아내기
    print("✅ 원천 데이터 로컬 압축 해제 완료!")
else:
    print(f"❌ 파일을 찾을 수 없습니다: {img_zip_path}")

# [기능 6] 라벨(JSON) 데이터 로컬 압축 해제 파이프라인
print("\n🚀 초고속 로컬 디스크로 라벨 데이터 압축 해제 중...")
if os.path.exists(label_zip_path):
    with zipfile.ZipFile(label_zip_path, 'r') as zip_ref:
        zip_ref.extractall(label_extract_path)
    print("✅ 라벨 데이터 로컬 압축 해제 완료!")
else:
    print(f"❌ 파일을 찾을 수 없습니다: {label_zip_path}")

print("\n🎉 로컬 I/O 최적화 세팅이 완벽하게 끝났습니다!")
```

```python
import os
import json

# [기능 1] 로컬 디스크의 라벨(JSON) 폴더 경로 설정
local_label_dir = '/content/AIHUB_Raw/labels'
print("🔍 로컬 디스크에서 JSON 파일 1개의 내부 구조를 분석합니다...\n")

found = False
# [기능 2] os.walk를 활용한 디렉토리 및 하위 폴더 순회 탐색 알고리즘
for root, dirs, files in os.walk(local_label_dir):
    for file in files:
        if file.endswith('.json'):
            sample_path = os.path.join(root, file)

            # [기능 3] JSON 파일 로드 및 내용 추출
            with open(sample_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                print(f"📄 찾은 파일명: {file}")
                print("--------------------------------------------------")
                # [기능 4] 가독성을 위한 JSON 구조 시각화 (첫 1000자 제한으로 메모리 보호)
                print(json.dumps(data, indent=4, ensure_ascii=False)[:1000])
                print("\n--------------------------------------------------")

            found = True
            break # 파일 1개만 확인하면 되므로 내부 루프 종료
    if found:
        break # 찾았으면 외부 루프도 즉시 종료

if not found:
    print("❌ JSON 파일을 찾지 못했습니다. 압축 경로를 다시 확인해주세요.")
```
